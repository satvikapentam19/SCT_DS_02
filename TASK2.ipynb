```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings("ignore")

sns.set(style="whitegrid")
```

```python
data_path = r"C:\Users\satvi\Downloads\titanic\train.csv"
df = pd.read_csv(data_path)

print("Dataset Info:")
print(df.info())

print("\nFirst 5 rows:")
print(df.head())
```

```python
print("\nMissing Values Summary:")
print(df.isnull().sum())

df = df.copy()

df['Age'] = df['Age'].fillna(df['Age'].median())

df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

df = df.drop(columns=['Cabin'])

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
```

```python
print("\nSummary Statistics:")
print(df.describe(include='all'))
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Survived', data=df, hue='Survived', palette='Set2', legend=False)
plt.title('Survival Count')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Sex', hue='Survived', data=df, palette='Set1')
plt.title('Survival by Gender')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Pclass', hue='Survived', data=df, palette='Set3')
plt.title('Survival by Passenger Class')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df[df['Survived']==1]['Age'], color='green', label='Survived', kde=True)
sns.histplot(df[df['Survived']==0]['Age'], color='red', label='Not Survived', kde=True)
plt.legend()
plt.title('Age Distribution by Survival')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Embarked', hue='Survived', data=df, palette='muted')
plt.title('Survival by Embarked Port')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='Pclass', y='Fare', data=df, palette='pastel')
plt.title('Fare Distribution by Passenger Class')
plt.show()
```
